```python
def depositar(saldo):
  print(f" seu saldo atual é R$ {saldo}")
  deposito = float(input("Informe o valor do depósito: "))
  saldo = saldo + deposito
  return saldo

def sacar(saldo):
  while True:
    print(f" Seu saldo atual é R$ {saldo}")
    saque = float(input("Informe o valor do saque: "))
    if saque > saldo:
      print("Saldo insuficiente, saque não realizado. ")
      opcao_saque = input("Deseja informar novo valor para saque? (sim) (não)")
      if opcao_saque.lower() != "sim":
        break
    else:
      saldo = saldo - saque
      break
  return saldo
```

```python
print("Bem vindo ao GEN-CAIXA")

saldo = float(input("Informe seu saldo "))
while True:
  print("Escolha uma das opções")
  print("1 - ver saldo")
  print("2 - depositar")
  print("3 - sacar")
  print("qualquer outra opção - sair")

  resposta = input(" Digite sua opção: ")

  if resposta =="1":
    print(f" Seu saldo é R$ {saldo:.2f}")
  elif resposta =="2":
    saldo = (depositar(saldo))
    print(f" Seu saldo atual é R$ {saldo:.2f}")
  elif resposta == "3":
    saldo = sacar(saldo)
    print(f" Seu saldo atual é R$ {saldo:.2f}")
  else:
    print("Sistema encerrado - Volte sempre ")
    break

```
